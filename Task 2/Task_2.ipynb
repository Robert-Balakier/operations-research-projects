## A. Identify the Optimization Problem.

Amazon wants to find the lowest cost way to move cargo from its hubs and focus cities to fulfillment centers. The solution must meet the demand of each fulfillment center while staying within the capacity limits of the hubs and focus cities. The goal is to determine how much cargo should be shipped on each route while minimizing total transportation costs (Western Governors University, n.d.-b).

## B. Create mathematical representations of the components of the optimization problem you identified.

### B1.

The objective is to minimize the total transportation cost.

Minimize Z = ΣᵢΣⱼ(cij × xij) + ΣᵢΣₖ(dik × yik) + ΣⱼΣₖ(ejk × zjk)

Where:

- xij = cargo shipped from hub i to focus city j
- yik = cargo shipped from hub i to fulfillment center k
- zjk = cargo shipped from focus city j to fulfillment center k
- cij = cost from hub i to focus city j
- dik = cost from hub i to fulfillment center k
- ejk = cost from focus city j to fulfillment center k

This objective function minimizes the cost of all cargo shipped through the network (Western Governors University, n.d.-a).

### B2.

Hub capacity:

Σⱼxij + Σₖyik ≤ Hi, for every hub i

Focus city capacity:

Σᵢxij ≤ Fj, for every focus city j

Flow balance:

Σₖzjk ≤ Σᵢxij, for every focus city j

Fulfillment center demand:

Σᵢyik + Σⱼzjk = Dk, for every fulfillment center k

Non-negativity:

xij ≥ 0, yik ≥ 0, and zjk ≥ 0 for all available routes


### B3.

The decision variables represent the amount of cargo shipped on each route.

- xij = cargo shipped from a hub to a focus city
- yik = cargo shipped from a hub to a fulfillment center
- zjk = cargo shipped from a focus city to a fulfillment center

The model calculates the best values for these variables to minimize shipping costs.

## D. Assess the Risks and Limitations Involved in Your Recommended Optimization Approach

1. One risk of this approach is that the model assumes shipping costs are constant, but fuel prices and transportation costs can change.

2. A second limitation is that the model assumes demand is known and stable, but fulfillment center demand can change over time.

3. A third limitation is that the model does not include unexpected disruptions such as weather, labor shortages, equipment failure, or transportation delays.

4. A fourth limitation is that the model assumes hub and focus city capacities are accurate, but real capacity may change.

5. A final limitation is that linear programming assumes the relationships are linear, while real world logistics may be more complex.

Because of these risks and limitations, the model should be used as a support tool rather than an exact prediction.
