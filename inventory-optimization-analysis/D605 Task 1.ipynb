### A. Identify a business need in the given scenario that could benefit from an optimization approach

GL Organics is experiencing inventory management issues that are causing overstocking, and unnecessary product waste. Because many of the company’s products are perishable, inefficient inventory management can lead to food waste, and lost sales opportunities.

1. For Scenario 3, GL Organics is experiencing inventory management issues resulting in overstocking, stockouts, wasted products. Due to the perishable nature of many of the company’s products, inefficient inventory management can lead to unnecessary food waste. The company would benefit from an optimization that improves inventory efficiency by better balancing product demand, warehouse capacity, and product shelf life. An improved inventory system could help identify products that are expiring, improve reorder timing, and reduce unnecessary waste.

2. This is a linear optimization problem because perishable items have fixed shelf lives, and variables such as inventory levels, waste quantities, reorder quantities, and warehouse capacity can all be modeled using linear equations and constraints.

3. The type of optimization problem identified is an **inventory optimization problem** because the company must balance inventory levels, product shelf life, warehouse capacity, and customer demand while minimizing waste.

### B. Identify the optimization objective, decision variables, and two or more constraints that are components of your optimization approach.

The optimization objective is to minimize inventory waste, and overall inventory costs.

The decision variables include reorder quantities, inventory levels for each product, reorder timing, and warehouse space allocation.

The constraints include limited warehouse capacity, product shelf life because of perishability and the need to meet customer demand forecasts.

1. Endpoint considerations include reducing unnecessary food waste, maintaining product freshness. The final optimization solution should also ensure that inventory levels remain within warehouse capacity limits.

### C. Recommend an optimization method that is appropriate for the problem you identified.

1. The recommended optimization method for this problem is **Linear Programming**. Linear Programming is appropriate because the company’s inventory management challenges can be modeled using linear objective functions, such as warehouse capacity, product shelf life limits, reorder quantities, and customer demand. This method can help GL Organics determine the optimal inventory levels and reorder schedules that minimize waste.
